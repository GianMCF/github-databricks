```python

from pyspark.sql import functions as F

tabla = "stasys_team2.gianmarco.consultas_denuncias_silver"

df = spark.table(tabla)

# Consulta 1: volumen por tipo de solicitud
resultado = (
    df.groupBy("DE_TIPO_SOLICITUD")
      .agg(F.count("*").alias("total_solicitudes"))
      .orderBy(F.desc("total_solicitudes"))
)

display(resultado)
```

```python

from pyspark.sql import functions as F

resultado_mensual = (
    df.withColumn(
        "fecha",
        F.to_date(F.col("FE_REGISTRO"))
    )
    .filter(F.col("fecha").isNotNull())
    .withColumn(
        "periodo",
        F.date_format("fecha", "yyyy-MM")
    )
    .groupBy("periodo")
    .agg(F.count("*").alias("total_solicitudes"))
    .orderBy("periodo")
)

display(resultado_mensual)
```
